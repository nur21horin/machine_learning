## Feature Engineering

```python
# Extract time-based features from 'Timestamp'
df['hour'] = df['Timestamp'].dt.hour
df['dayofweek'] = df['Timestamp'].dt.dayofweek
df['dayofmonth'] = df['Timestamp'].dt.day
df['month'] = df['Timestamp'].dt.month
df['year'] = df['Timestamp'].dt.year
df['weekofyear'] = df['Timestamp'].dt.isocalendar().week.astype(int)
df['is_weekend'] = (df['Timestamp'].dt.dayofweek >= 5).astype(int)
df['is_month_start'] = df['Timestamp'].dt.is_month_start.astype(int)
df['is_month_end'] = df['Timestamp'].dt.is_month_end.astype(int)
df['is_quarter_start'] = df['Timestamp'].dt.is_quarter_start.astype(int)
df['is_quarter_end'] = df['Timestamp'].dt.is_quarter_end.astype(int)
df['is_year_start'] = df['Timestamp'].dt.is_year_start.astype(int)
df['is_year_end'] = df['Timestamp'].dt.is_year_end.astype(int)

display(df.head())
```

### Encoding Categorical Features

We need to convert the categorical columns ('Pump Status' and 'Anomaly Detected') into a numerical format that machine learning models can understand.

*   **'Pump Status'**: This is a nominal categorical variable (ON/OFF). One-hot encoding is suitable here to avoid implying any ordinal relationship.
*   **'Anomaly Detected'**: This is a binary target variable ('Yes'/'No'). We can map 'Yes' to 1 and 'No' to 0.

```python
# One-hot encode 'Pump Status'
df = pd.get_dummies(df, columns=['Pump Status'], drop_first=True)

# Map 'Anomaly Detected' to numerical (0 and 1)
df['Anomaly Detected'] = df['Anomaly Detected'].map({'Yes': 1, 'No': 0})

# Display the head and info of the DataFrame to see the changes
display(df.head())
display(df.info())
```

## Data Preparation for Modeling: SMOTE and Class Weighting

To prepare the data for modeling, especially for imbalanced datasets, we will combine SMOTE (Synthetic Minority Oversampling Technique) with cost-sensitive learning through class weights.

First, we need to re-split the `df` DataFrame into `train`, `val`, and `test` sets to ensure that all engineered features and encoded columns are present in these subsets. Then, we will apply SMOTE to the training data to balance the classes and calculate class weights for use in the model.

```python
# Re-split data to include all engineered features and encoded columns
train_size = int(0.8 * len(df))
val_size = int(0.1 * len(df))

train_df = df.iloc[:train_size].copy()
val_df = df.iloc[train_size : train_size + val_size].copy()
test_df = df.iloc[train_size + val_size :].copy()

print(f"Original training set shape: {train_df.shape}")
print(f"Original validation set shape: {val_df.shape}")
print(f"Original test set shape: {test_df.shape}")
```

### Separate Features and Target

Now, let's define our features (X) and the target variable (y) for the training, validation, and test sets. We'll drop the 'Timestamp' column as its information has been captured in the extracted time-based features, and the 'Anomaly Detected' column is our target.

```python
# Define features (X) and target (y)
TARGET_COLUMN = 'Anomaly Detected'
FEATURES = [col for col in df.columns if col not in ['Timestamp', TARGET_COLUMN]]

X_train = train_df[FEATURES]
y_train = train_df[TARGET_COLUMN]

X_val = val_df[FEATURES]
y_val = val_df[TARGET_COLUMN]

X_test = test_df[FEATURES]
y_test = test_df[TARGET_COLUMN]

print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_val shape: {X_val.shape}, y_val shape: {y_val.shape}")
print(f"X_test shape: {X_test.shape}, y_test shape: {y_test.shape}")

# Check initial class distribution in y_train
print("\nInitial class distribution in training data (y_train):")
display(y_train.value_counts())
```

### Apply SMOTE to Training Data

We will use SMOTE to oversample the minority class ('Anomaly Detected' = 1) in the training data to address the class imbalance. Based on the request, we will aim for a specific ratio (e.g., 1:2 for minority to majority).

```python
from imblearn.over_sampling import SMOTE

# Instantiate SMOTE
# Calculate the current ratio of minority (1) to majority (0)
minority_count = y_train.value_counts()[1]
majority_count = y_train.value_counts()[0]

# Target a 1:2 ratio (minority:majority)
sampling_strategy_ratio = {0: majority_count, 1: int(majority_count / 2)}

sm = SMOTE(sampling_strategy=sampling_strategy_ratio, random_state=42)

# Apply SMOTE only to the training data
X_train_resampled, y_train_resampled = sm.fit_resample(X_train, y_train)

print("\nClass distribution after SMOTE:")
display(y_train_resampled.value_counts())
print(f"Resampled X_train shape: {X_train_resampled.shape}, resampled y_train shape: {y_train_resampled.shape}")
```

### Calculate Class Weights

To further address the imbalance and introduce cost-sensitive learning, we will calculate class weights. These weights can be passed to many machine learning models to give more importance to the minority class during training. We calculate these weights based on the *resampled* training data.

```python
from sklearn.utils import class_weight

# Calculate class weights for the resampled data
class_weights = class_weight.compute_class_weight(
    class_weight='balanced',
    classes=np.unique(y_train_resampled),
    y=y_train_resampled
)

# Convert to a dictionary for easier use in models
class_weights_dict = dict(zip(np.unique(y_train_resampled), class_weights))

print("Class weights:")
print(class_weights_dict)
```

## Model Training and Evaluation

Now that the data is prepared, let's train different classification models. We will evaluate each model on both the original (imbalanced) training data and the SMOTE-resampled training data to understand the impact of balancing techniques. We'll use the validation set (`X_val`, `y_val`) for evaluation.

### 1. Random Forest Classifier

We will train a Random Forest Classifier using `class_weight='balanced'` to handle class imbalance, first on the original training data and then on the SMOTE-resampled data.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, classification_report, confusion_matrix

def train_evaluate_model(model, X_train_data, y_train_data, X_val, y_val, model_name, data_source):
    """Trains and evaluates a given model."""
    print(f"\n--- Training {model_name} on {data_source} Data ---")

    # Train the model
    model.fit(X_train_data, y_train_data)

    # Make predictions on the validation set
    y_pred = model.predict(X_val)
    y_proba = model.predict_proba(X_val)[:, 1] # Probability of the positive class

    # Evaluate the model
    print(f"\n{model_name} - {data_source} Evaluation on Validation Set:")
    print(f"Accuracy: {accuracy_score(y_val, y_pred):.4f}")
    print(f"Precision: {precision_score(y_val, y_pred):.4f}")
    print(f"Recall: {recall_score(y_val, y_pred):.4f}")
    print(f"F1-Score: {f1_score(y_val, y_pred):.4f}")
    print(f"ROC AUC Score: {roc_auc_score(y_val, y_proba):.4f}")
    print("\nClassification Report:")
    print(classification_report(y_val, y_pred))
    print("\nConfusion Matrix:")
    print(confusion_matrix(y_val, y_pred))

# -----------------------------------------------------------------------------
# 1. Random Forest Classifier on Original Imbalanced Data
print("\n### Random Forest on Original Data")
rf_model_original = RandomForestClassifier(random_state=42, class_weight='balanced')
train_evaluate_model(rf_model_original, X_train, y_train, X_val, y_val, 'Random Forest', 'Original')

# -----------------------------------------------------------------------------
# 2. Random Forest Classifier on SMOTE-Resampled Data
print("\n### Random Forest on SMOTE-Resampled Data")
rf_model_smote = RandomForestClassifier(random_state=42, class_weight='balanced')
train_evaluate_model(rf_model_smote, X_train_resampled, y_train_resampled, X_val, y_val, 'Random Forest', 'SMOTE-Resampled')
```
