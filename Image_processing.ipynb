```python
import numpy as np
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
```

```python
(x_train,y_train),(x_test,y_test)=mnist.load_data()
```

```python
x_train=x_train.astype('float32')/255.0
x_test=x_test.astype('float32')/255.0
```

```python
x_train=np.expand_dims(x_train,axis=-1)
x_test=np.expand_dims(x_test,axis=-1)
```

```python
print(f"Reshaped Trainig Data Shape:{x_train.shape}")
print(f"Reshaped Testing Data Shape:{x_test.shape}")
```

```python
num_classes=10
y_train=to_categorical(y_train,num_classes)
y_test=to_categorical(y_test,num_classes)
```

```python
print(f"One Hot Encoded Training Data Shape:{y_train.shape}")
print(f"One Hot Encoded Testing Data Shape:{y_test.shape}")
```
