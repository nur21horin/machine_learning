```python
import scipy
from scipy import stats
```

```python
data=[10,12,12,43,23,45,34]
print(stats.skew(data))
print(stats.kurtosis(data))
```

```python
from scipy import stats

data = [10, 12, 15, 20, 25]

print(stats.mean(data))
```

```python
from scipy.stats import ttest_1samp

data = [10, 12, 15, 20, 25]

result = ttest_1samp(data, 15)

print(result)
```

```python
from scipy.spatial.distance import euclidean

a = [1, 2]
b = [4, 6]

print(euclidean(a, b))
```

```python
from scipy.optimize import minimize

def f(x):
    return x**2 + 5*x + 6

result = minimize(f, x0=0)

print(result.x)
```

```python
from scipy.stats import norm

print(norm.pdf(0))
print(norm.cdf(1.96))
```
